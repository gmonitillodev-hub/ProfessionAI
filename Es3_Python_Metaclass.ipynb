```python
import time


def measure_execution_time(func):
    # TODO: Implementare funzione per tracciare l'esecuzione
    def wrapper(*args, **kwargs):
      start_time = time.time()
      result = func(*args, **kwargs) # Pass arguments to the original function
      end_time = time.time()

      print(f"Tempo di esecuzione -> {end_time - start_time}")
      return result # Return the result of the original function
    return wrapper


class UpperAttrMeta(type):
     def __new__(cls, name, bases, dct):
        # 1. Creiamo prima la classe usando il super()
        nuova_classe = super().__new__(cls, name, bases, dct)

        # 2. Ora la classe esiste! Possiamo usare hasattr, getattr, setattr e delattr
        if hasattr(nuova_classe, "foo"):
            valore = getattr(nuova_classe, "foo")

            # Crea l'attributo in maiuscolo
            setattr(nuova_classe, "FOO", valore)

            # Elimina l'attributo originale in minuscolo
            delattr(nuova_classe, "foo")

        # 3. Restituiamo la classe modificata
        return nuova_classe


class MyClass(metaclass=UpperAttrMeta):
    foo = "bar"
    baz = "qux"


@measure_execution_time
def calculate_power(numbers):
    tot = 1

    for x in range(numbers):
        tot += x ** 10

    return tot


 # BONUS:

def cache_with_expiry(expiry_time):
    cache = {} # Cache should be here, outside the wrapper
    def decorator(func):
        def wrapper(*args, **kwargs):
            # Create a unique key for the cache based on arguments
            key = (args, frozenset(kwargs.items()))

            if key in cache and (time.time() - cache[key]['timestamp']) < expiry_time:
                print("Retrieving result from cache...")
                return cache[key]['result']
            else:
                result = func(*args, **kwargs) # Call the original function with its arguments
                cache[key] = {'result': result, 'timestamp': time.time()}
                return result
        return wrapper
    return decorator

class SingletonMeta(type):
    # TODO: Implementare metaclasse che permette pattern Singleton
    _instances = {}
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class SingletonClass(metaclass=SingletonMeta):
    pass


@cache_with_expiry(expiry_time=5)  # Cache expiry time set to 5 seconds
def calculate_multiply(x, y):
    print("Calculating product of two numbers...")
    return x * y


# PROGRAMMA

## ESERCIZIO BASE ##

# Tempo di esecuzione
result = calculate_power(10)
print("Result:", result)

# OUTPUT PREVISTO:
# function calculate_multiply took 0.1094448566 seconds to execute --> I secondi variano da pc a pc
# Result: 90908590909924242424241424242424243424242424241924242424242500001

# Metaclasse attributi upper case
print(hasattr(MyClass, "foo" ))
# False, perché attributi sono UPPERCASE

print(hasattr(MyClass, "FOO" ))
# True, perché foo è diventato 'FOO'

print(MyClass.FOO)  # Accediamo all'attributo in Uppercase


## BONUS ##

## SINGLETON TEST ##
instance1 = SingletonClass()
instance2 = SingletonClass()

# Check if both instances refer to the same object
print(f"Are the same obj ? {instance1 is instance2}")

# OUTPUT PREVISTO: True


## CACHE RESULT TEST ##
print(calculate_multiply(23, 5))
print(calculate_multiply(23, 5))

time.sleep(2)

print(calculate_multiply(23, 5))
```
