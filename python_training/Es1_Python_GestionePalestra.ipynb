```python
prenotazioni = [
    {"nome": "Alice", "attività": "Yoga"},
    {"nome": "Luca", "attività": "CrossFit"},
    {"nome": "Marco", "attività": "Yoga"},
    {"nome": "Elena", "attività": "Pilates"},
    {"nome": "Davide", "attività": "CrossFit"},
    {"nome": "Sofia", "attività": "Yoga"},
    {"nome": "Francesco", "attività": "CrossFit"},
    {"nome": "Giulia", "attività": "Pilates"},
    {"nome": "Andrea", "attività": "Yoga"},
    {"nome": "Giorgia", "attività": "Boxe"},
]

print(prenotazioni)

catalogo_attivita = ({"nome":"Yoga", "posti_massimi": 5},{"nome": "CrossFit", "posti_massimi": 15},{"nome": "Pilates", "posti_massimi": 3}, {"nome":"Boxe", "posti_massimi": 35})
print(catalogo_attivita)


def conta_iscrizioni_per_attivita(catalogo, prenotazioni):
    iscrizioni_per_attivita = {}
    for attivita in catalogo:
        nome_attivita = attivita["nome"]
        # Count reservations where 'attività' matches 'nome_attivita'
        count = len([p for p in prenotazioni if p["attività"] == nome_attivita])
        iscrizioni_per_attivita[nome_attivita] = count
    return iscrizioni_per_attivita

# Example usage of the function
iscrizioni = conta_iscrizioni_per_attivita(catalogo_attivita, prenotazioni)
print("Iscrizioni per attività:", iscrizioni)
```

```python
def posti_disponibili_per_attivita(catalogo, prenotazioni):
    posti_disponibili = {}

    iscrizioni_per_attivita = conta_iscrizioni_per_attivita(catalogo, prenotazioni)
    for attivita in catalogo:
        nome_attivita = attivita["nome"]
        posti_disponibili[nome_attivita] = attivita["posti_massimi"] - iscrizioni_per_attivita.get(nome_attivita, 0)

    return posti_disponibili

print(posti_disponibili_per_attivita(catalogo_attivita, prenotazioni))
```

```python
def filtra_per_attività(prenotazioni, filters):

  prenotazioni_filtered = []

  filter_value = filters["attività"]
  if not filter_value:
    print("Filtro non valido")
    return prenotazioni

  for prenotazione in prenotazioni:
    if prenotazione["attività"] == filter_value:
      prenotazioni_filtered.append(prenotazione)

  return prenotazioni_filtered


def filtra_prenotazioni(function, prenotazioni, filters):
  return function(prenotazioni, filters)

prenotazioni_filtered = filtra_prenotazioni(filtra_per_attività, prenotazioni, {"attività": "Yoga"})
print(prenotazioni_filtered)
```

```python
nomi_partecipanti =  map(lambda pre: pre["nome"], prenotazioni)
print(list(nomi_partecipanti))

iscrizioni_pilates = filter(lambda pil: pil["attività"] == "Pilates", prenotazioni)
print(list(iscrizioni_pilates))
from functools import reduce

iscrizioni_totali = reduce(lambda acc, pre: acc + 1, prenotazioni, 0)
print(iscrizioni_totali)

sorted_list = sorted(prenotazioni, key=lambda pre: pre["attività"])
print(sorted_list)

```
