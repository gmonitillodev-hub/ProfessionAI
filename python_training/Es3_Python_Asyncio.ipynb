```python
import asyncio

async def download_img(name, dow_time):
  print(f"Inizio download: {name}")

  await asyncio.sleep(dow_time)

  print(f"Fine download: {name}")

  return True

async def main():

  ''' Non è multithreading, ma gestione a singolo thread simulato '''
  result = await asyncio.gather(
    download_img("img1", 3),
    download_img("img2", 5),
    download_img("img3", 1)
  )

  print(result)


await main()
```
